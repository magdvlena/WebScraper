```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("my_app").master("local[*]").getOrCreate()
```

* appname - nazwa aplikacji
* aplikacja - program który korzysta ze sparka; ały proces uruchomiony na klastrze/lokalnie z przydzielonymi zasobami CPU/RAM; stworzona aby pracować z bardzo dużymi zbiorami danych
* sesja - punkt wejścia w kodzie; pozwala programowi rozmawiać z silnikiem sparka; pulpit sterowniczy; obiekt pythonowy przez który wydaję polecenia; Służy do pracy z ramkami danych (DataFrame), tabele SQL i zapytaniami
* master local - określenie trybu uruchomienia; u nas lokalnie, wykorzystując wszystkie możliwe rdzenie  procesora
* get or create - metoda służy do pobierania istniejącej sesji Sparka lub tworzenia nowej, jeśli żadna jeszcze nie działa w danym procesie. Jeśli taka sesja już istnieje, zignoruje nowo przekazane parametry (np. inną nazwę appName) i po prostu zwróci referencję do tej działającej sesji. Jeśli nie ma aktywnej sesji, Spark stworzy nowy obiekt SparkSession na podstawie podanej konfiguracji. W ramach jednego procesu Pythona w PySparku może działać tylko jeden kontekst wykonawczy (SparkContext). Gdyby stworzyć osobną sesję bez tego mechanizmu, mogłoby dojść do konfliktu zasobów.
* SparkContext - silnik pod spodem (kontekst wykonawczy). Zarządza fizycznym wykonaniem zadań, alokacją pamięci i komunikacją z procesami roboczymi.

* SparkContext: Niskopoziomowy silnik wykonawczy znajdujący się wewnątrz sesji, który odpowiada za nawiązywanie połączenia z klastrem i zarządzanie zasobami.

* Operator is w Pythonie sprawdza tożsamość obiektów w pamięci RAM (to znaczy, czy obie zmienne celują dokładnie w ten sam adres w pamięci).
* Przypisanie nowej zmiennej nie tworzy nowego silnika ani osobnego połączenia, a jedynie nową referencję (wskaźnik) do tej samej, istniejącej sesji.

Metoda SparkSession.getActiveSession() służy do pobierania referencji do aktualnie uruchomionej i aktywnej sesji Sparka w ramach danego procesu Pythona.
