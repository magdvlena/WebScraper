```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("my_app").master("local[*]").getOrCreate()
```

aplikacja spark - czyli uruchomiony program ktory wykorzystuje pachispark
spark jest stworzony zeby pracowac z bardzo duzymi zbiorami danych


```python
spark
```

spark ui - graficzny interfejs ktory umozliwa monitorowanie i wykorzystywanie aplikacji spark np wydajnosci; on dziala wewnatrz zdalneog srodowiska wykonawczego; klikajac otrzymamy błąd
mamy podana wersje
local
app name - nadana przez nas nazwa

```python
data = [(1, "Ala"), (2, "Jan"), (3, "Ola")]
```

```python
columns = ["id", "name"]
```

```python
df = spark.createDataFrame(data, columns)
```

```python
df.show()
```
